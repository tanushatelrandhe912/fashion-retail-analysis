```python
i = 'Discount_Price'
```

```python
q1 = np.percentile(df[i],25)
q3 = np.percentile(df[i],75)

iqr = q3-q1

print(q1)
print(q3)
print(iqr)
```

```python
c1 = q1-1.5*(iqr)
c2 = q3+1.5*(iqr)

t = df[(df[i]>c2) | (df[i]<c1)]
t=t.index
df=df.drop(labels=t)
```

```python
t= df[(df[i]>c2)|(df[i]<c1)]
t
```
